```python
dict_list = [{'Name':'Tom'},{'Name':'Jane','Sales':300, 'Job Title':'Sales'},{'Name':'Jane','Sales':300, 'Job Title':'Sales'},{'Name':'Steve','Sales':350},{'Name':'Lucy','Sales':400,'Job Title':'Sales', 'Location':'London'}]
df = pd.DataFrame(dict_list)
df
```

```python
df.isnull().sum()
```

```python
#removes rows with too many null na values 
#this doesn't chnage teh data frame its just a view method 
df.dropna(thresh=3)
```

```python
df.fillna(0)
```

```python
df['Sales'].fillna(0)
```

```python
df['Filled_sales'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df['Sales'].mean()
```
