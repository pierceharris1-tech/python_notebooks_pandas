```python
fmri = sns.load_dataset('fmri')
fmri.head()
```

```python
sns.lineplot(x="timepoint", y="signal", data=fmri)
plt.title("Average FMRI across time")
plt.xlabel("Timepoint")
plt.ylabel("signal")
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

## Shows the Average 

```python
sns.barplot(x="day", y="total_bill", data=tips)
```

## Just shows the highest bill per day 

### All values for sunday are on sunday but you can only see the largest one as it covers the other values 

```python
plt.bar(tips.day, tips.total_bill)
```

```python
sns.catplot(x='day', y="total_bill", errorbar=None, data=tips, col="time", hue="sex", kind= "bar")
```

```python
sns.barplot(x='day', y="total_bill", errorbar=None, data=tips, hue="sex")
```
